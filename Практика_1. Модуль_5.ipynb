```python
import pandas as pd
import numpy as np
np.random.seed(0)
countries = np.random.choice(["Россия", "Индонезия", "Индия", "Катар", "Аргентина", "Австралия", "Франция", "ЮАР", "Танзания",
 "США", "Испания", "Кения", "Канада", "Бразилия", "Перу", "Чили", "Алжир", "Мексика", "Саудовская Аравия", 
"Италия", "Марокко", "Норвегия", "Малайзия", "Йемен", "Мозамбик", "Эфиопия", "Судан", "Иран", "Китай", "Мали"], size = 30)
religion = np.random.choice (["Христианство", "Ислам", "Буддизм", "Иудаизм", "Буддизм"], 30, replace=True)
political = ["Демократия", "Авторитаризм"] * 15
continent = np.random.choice (["Африка", "Австралия", "Южная Америка", "Северная Америка", "Евразия"], 30, replace = True)
place_in_the_world_by_area = np.random.choice(range(200), 30, replace=False)
VVP = np.random.choice ([60000, 50000, 40000, 30000, 20000, 10000, 5000, 2500, 1000, 500], 30, replace = True)
place_population = np.random.choice(range(200), 30, replace=False)
year = [1700, 1900, 1600, 1200, 1300, 1100] * 5

data = {
    "countries": countries,
    "religion": religion,
    "political": political,
    "continent": continent,
    "place_in_the_world_by_area": place_in_the_world_by_area,
    "VVP": VVP,
    "place_population": place_population,
    "year": year}

df = pd.DataFrame(data)

display(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Создаем перекрестную таблицу
cross_tab = pd.crosstab(df['continent'], df['religion'])

# Столбчатая диаграмма
plt.figure(figsize=(12, 6))
cross_tab.plot(kind='bar', stacked=False)
plt.title('Распределение религий по континентам')
plt.xlabel('Континент')
plt.ylabel('Количество стран')
plt.xticks(rotation=45)
plt.legend(title='Религия')
plt.tight_layout()
plt.show()
```
