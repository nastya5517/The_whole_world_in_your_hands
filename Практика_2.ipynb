#Задача №9

```python
import numpy as np
m = np.random.rand(3, 4)
print("Матрица:", m)
def matrix_to_vector(m):
    rows = len(m)
    cols = len(m)
    vector = []
    for n in range(rows):
        for  v in range(cols):
            vector.append(m[n][v])
    return vector
vector = matrix_to_vector(m)
print("Вектор:", vector)
```
